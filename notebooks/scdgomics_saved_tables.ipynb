```python
# Import libs
import pandas as pd 
import numpy as np
import scanpy as sc
import os, sys
import csv
```

```python
# Load genes (GO term: "cell part morphogenesis")
hNcx_vs_mNcx = pd.read_csv("/group/kalebic/Nikola/1_single_cell_final/Nikola_final/final/scdgomics/tables/cell_part_morphogenesis_hNcx_vs_mNcx.txt", header = None).rename(columns={0:"genes"})['genes'].tolist()
mDG_vs_mNcx = pd.read_csv("/group/kalebic/Nikola/1_single_cell_final/Nikola_final/final/scdgomics/tables/cell_part_morphogenesis_mDG_vs_mNcx.txt", header = None).rename(columns={0:"genes"})['genes'].tolist()
```

```python
# Print
print(len(hNcx_vs_mNcx))
print(len(mDG_vs_mNcx))
```

```python
# Intersect
intersected = set(hNcx_vs_mNcx).intersection(mDG_vs_mNcx)
print(len(intersected))
```

```python
# Convert to a df 
intersected_df = pd.DataFrame(intersected)
intersected_df = intersected_df.rename(columns={0:"genes"})
intersected_df
```

```python
# Save
intersected_df['genes'] = intersected_df['genes'].astype(str)
intersected_df.to_csv("/group/kalebic/Nikola/1_single_cell_final/Nikola_final/final/scdgomics/tables/Intersected_degs_cell_part_morphogenesis.csv", index=False, quoting=csv.QUOTE_NONNUMERIC)

with pd.ExcelWriter("/group/kalebic/Nikola/1_single_cell_final/Nikola_final/final/scdgomics/tables/Intersected_degs_cell_part_morphogenesis.xlsx") as writer:
    intersected_df.to_excel(writer, sheet_name='genes', index=False)
```
